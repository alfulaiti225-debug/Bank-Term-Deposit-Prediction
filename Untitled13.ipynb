```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check unknown values in categorical columns
print("\nUnknown Values:")
categorical_cols = df.select_dtypes(include=['object', 'str']).columns

for col in categorical_cols:
    unknown_count = (df[col] == 'unknown').sum()
    print(col, ":", unknown_count)
```

```python

# Create a copy of the original dataset
df_clean = df.copy()

# Replace unknown values with the most frequent category
for col in categorical_cols:
    known_values = df_clean.loc[
        df_clean[col] != 'unknown', col
    ]

    if (df_clean[col] == 'unknown').any():
        most_frequent = known_values.mode()[0]
        df_clean[col] = df_clean[col].replace(
            'unknown', most_frequent
        )

# Check unknown values after preprocessing
print("Unknown Values After Preprocessing:")

for col in categorical_cols:
    print(col, ":", (df_clean[col] == 'unknown').sum())

print("\nDataset Shape:", df_clean.shape)
```

```python
# Remove call duration to avoid data leakage
df_clean = df_clean.drop(columns=['duration'])

# Check dataset shape after feature removal
print("Dataset Shape After Feature Removal:", df_clean.shape)

# Check whether duration was removed
print("Duration Column Exists:", 'duration' in df_clean.columns)
```

```python
# Check duplicate records
duplicate_count = df_clean.duplicated().sum()

print("Number of Duplicate Records:", duplicate_count)
print("Dataset Shape Before Duplicate Removal:", df_clean.shape)
```

```python
# Detect age outliers using IQR
Q1 = df_clean['age'].quantile(0.25)
Q3 = df_clean['age'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

age_outliers = df_clean[
    (df_clean['age'] < lower_bound) |
    (df_clean['age'] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Age Outliers:", len(age_outliers))
```

```python
# Separate features and target variable
X = df_clean.drop(columns=['y'])
y = df_clean['y'].map({'no': 0, 'yes': 1})

# Apply One-Hot Encoding to categorical features
X_encoded = pd.get_dummies(
    X,
    columns=X.select_dtypes(include=['object', 'str']).columns,
    dtype=int
)

print("Features Before Encoding:", X.shape)
print("Features After Encoding:", X_encoded.shape)

print("\nTarget Values:")
print(y.value_counts())

# Display encoded data sample
display(X_encoded.head())
```

```python
# Apply Standard Scaling
from sklearn.preprocessing import StandardScaler

# Create a copy of encoded features
X_scaled = X_encoded.copy()

# Select original numerical features
numeric_columns = X.select_dtypes(
    include=['int64', 'float64']
).columns

# Apply StandardScaler
scaler = StandardScaler()

X_scaled[numeric_columns] = scaler.fit_transform(
    X_encoded[numeric_columns]
)

# Compare before and after scaling
print("Before Scaling:")
display(X_encoded[numeric_columns].head())

print("\nAfter Scaling:")
display(X_scaled[numeric_columns].head())

print("\nDataset Shape:", X_scaled.shape)
```

```python
# Import train-test split
from sklearn.model_selection import train_test_split

# Use the original dataset before imputation and scaling
X_original = df.drop(columns=['y', 'duration'])
y_original = df['y'].map({'no': 0, 'yes': 1})

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_original,
    y_original,
    test_size=0.20,
    random_state=42,
    stratify=y_original
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

```python
# Compare two methods for handling unknown values

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

# Identify numerical and categorical columns
num_cols = X_train.select_dtypes(include='number').columns
cat_cols = X_train.select_dtypes(exclude='number').columns

# Method 1: Replace unknown with the most frequent category
X_train_mode = X_train.replace('unknown', np.nan)
X_test_mode = X_test.replace('unknown', np.nan)

mode_preprocessor = ColumnTransformer([
    ('num', StandardScaler(), num_cols),
    ('cat', Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(handle_unknown='ignore'))
    ]), cat_cols)
])

mode_model = Pipeline([
    ('preprocessing', mode_preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])

mode_model.fit(X_train_mode, y_train)
mode_predictions = mode_model.predict(X_test_mode)

# Method 2: Keep unknown as a separate category
keep_preprocessor = ColumnTransformer([
    ('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
])

keep_model = Pipeline([
    ('preprocessing', keep_preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])

keep_model.fit(X_train, y_train)
keep_predictions = keep_model.predict(X_test)

# Compare model performance
print("Method 1 - Mode Imputation")
print("Accuracy:", round(accuracy_score(y_test, mode_predictions), 4))
print("F1-score:", round(f1_score(y_test, mode_predictions), 4))

print("\nMethod 2 - Keep Unknown")
print("Accuracy:", round(accuracy_score(y_test, keep_predictions), 4))
print("F1-score:", round(f1_score(y_test, keep_predictions), 4))
```

```python
# Compare StandardScaler and MinMaxScaler
from sklearn.preprocessing import MinMaxScaler

scaling_results = []

for scaling_name, scaling_method in [
    ('StandardScaler', StandardScaler()),
    ('MinMaxScaler', MinMaxScaler())
]:

    preprocessor = ColumnTransformer([
        ('num', scaling_method, num_cols),
        ('cat', Pipeline([
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('encoder', OneHotEncoder(handle_unknown='ignore'))
        ]), cat_cols)
    ])

    model = Pipeline([
        ('preprocessing', preprocessor),
        ('classifier', LogisticRegression(max_iter=1000))
    ])

    model.fit(X_train_mode, y_train)
    predictions = model.predict(X_test_mode)

    scaling_results.append({
        'Scaling Method': scaling_name,
        'Accuracy': round(accuracy_score(y_test, predictions), 4),
        'F1-score': round(f1_score(y_test, predictions), 4)
    })

# Display comparison table
scaling_results_df = pd.DataFrame(scaling_results)

display(scaling_results_df)
```

```python

# Final preprocessing comparison table
comparison_df = pd.DataFrame({
    'Preprocessing Method': [
        'Mode + StandardScaler',
        'Keep Unknown + StandardScaler',
        'Mode + MinMaxScaler'
    ],
    'Accuracy': [
        accuracy_score(y_test, mode_predictions),
        accuracy_score(y_test, keep_predictions),
        scaling_results_df.loc[1, 'Accuracy']
    ],
    'F1-score': [
        f1_score(y_test, mode_predictions),
        f1_score(y_test, keep_predictions),
        scaling_results_df.loc[1, 'F1-score']
    ]
})

comparison_df[['Accuracy', 'F1-score']] = (
    comparison_df[['Accuracy', 'F1-score']].round(4)
)

display(comparison_df)
```

```python
# Compare preprocessing methods using cross-validation
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = []

for name, scaler_method in [
    ('StandardScaler', StandardScaler()),
    ('MinMaxScaler', MinMaxScaler())
]:
    preprocessor = ColumnTransformer([
        ('num', scaler_method, num_cols),
        ('cat', Pipeline([
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('encoder', OneHotEncoder(handle_unknown='ignore'))
        ]), cat_cols)
    ])

    model = Pipeline([
        ('preprocessing', preprocessor),
        ('classifier', LogisticRegression(max_iter=1000))
    ])

    scores = cross_val_score(
        model,
        X_train_mode,
        y_train,
        cv=cv,
        scoring='f1'
    )

    cv_results.append({
        'Method': name,
        'Mean F1-score': round(scores.mean(), 4),
        'F1 Std': round(scores.std(), 4)
    })

display(pd.DataFrame(cv_results))
```

```python

# Compare unknown handling methods using cross-validation

unknown_cv_results = []

for name, preprocessing, data in [
    ('Mode Imputation', mode_preprocessor, X_train_mode),
    ('Keep Unknown', keep_preprocessor, X_train)
]:
    model = Pipeline([
        ('preprocessing', preprocessing),
        ('classifier', LogisticRegression(max_iter=1000))
    ])

    scores = cross_val_score(
        model,
        data,
        y_train,
        cv=cv,
        scoring='f1'
    )

    unknown_cv_results.append({
        'Unknown Handling Method': name,
        'Mean F1-score': round(scores.mean(), 4),
        'F1 Std': round(scores.std(), 4)
    })

# Display comparison results
unknown_cv_df = pd.DataFrame(unknown_cv_results)

display(unknown_cv_df)
```

```python
# Prepare data for K-Means Clustering

from sklearn.preprocessing import StandardScaler

# Use original features without target and duration
X_cluster = df.drop(columns=['y', 'duration']).copy()

# Replace unknown values with the most frequent category
for col in X_cluster.select_dtypes(
    include=['object', 'str']
).columns:

    known_values = X_cluster.loc[
        X_cluster[col] != 'unknown', col
    ]

    if (X_cluster[col] == 'unknown').any():
        X_cluster[col] = X_cluster[col].replace(
            'unknown', known_values.mode()[0]
        )

# Convert categorical variables into numerical values
X_cluster = pd.get_dummies(X_cluster, dtype=int)

# Standardize the features
cluster_scaler = StandardScaler()
X_cluster_scaled = cluster_scaler.fit_transform(X_cluster)

# Display dataset information
print("Clustering Data Shape:", X_cluster_scaled.shape)
print("Missing Values:", X_cluster.isnull().sum().sum())
print("Data Ready for K-Means Clustering!")
```

```python
# Find the optimal number of clusters using Elbow Method

from sklearn.cluster import MiniBatchKMeans
import matplotlib.pyplot as plt
import numpy as np

# Select a sample for faster analysis
rng = np.random.RandomState(42)
sample_indices = rng.choice(
    X_cluster_scaled.shape[0],
    size=5000,
    replace=False
)

X_cluster_sample = X_cluster_scaled[sample_indices]

# Calculate inertia for different numbers of clusters
inertia_values = []

for k in range(2, 11):
    kmeans = MiniBatchKMeans(
        n_clusters=k,
        random_state=42,
        batch_size=1024,
        n_init=3
    )

    kmeans.fit(X_cluster_sample)
    inertia_values.append(kmeans.inertia_)

# Plot the Elbow Method
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    inertia_values,
    marker='o'
)

plt.title('Elbow Method for Customer Segmentation')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')

plt.xticks(range(2, 11))
plt.grid(True)

plt.show()
```

```python

# Correct Elbow Method using KMeans

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Calculate inertia for K = 2 to 10
inertia_kmeans = []

for k in range(2, 11):
    model_kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20,
        max_iter=300
    )

    model_kmeans.fit(X_cluster_sample)
    inertia_kmeans.append(model_kmeans.inertia_)

# Display inertia values
print("KMeans Inertia Values:")
for k, value in zip(range(2, 11), inertia_kmeans):
    print(f"K={k}: {value:.2f}")

# Plot the Elbow Method
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    inertia_kmeans,
    marker='o'
)

plt.title('Elbow Method for Customer Segmentation')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.xticks(range(2, 11))
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python

# Calculate Silhouette Scores for K-Means
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans

silhouette_results = []

for k in range(2, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_cluster_sample)

    score = silhouette_score(
        X_cluster_sample,
        labels,
        sample_size=2000,
        random_state=42
    )

    silhouette_results.append(score)

# Display Silhouette Scores
for k, score in zip(range(2, 11), silhouette_results):
    print(f"K={k}: Silhouette Score={score:.4f}")

# Plot Silhouette Scores
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_results,
    marker='o'
)

plt.title('Silhouette Score for Customer Segmentation')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.xticks(range(2, 11))
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python

# Apply K-Means Clustering with K=2
from sklearn.cluster import KMeans

final_kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

# Fit K-Means and predict clusters
cluster_labels = final_kmeans.fit_predict(X_cluster_scaled)

# Add cluster labels to the original dataset
df_clusters = df.copy()
df_clusters['Cluster'] = cluster_labels

# Display cluster sizes
print("Number of Customers in Each Cluster:")
print(df_clusters['Cluster'].value_counts().sort_index())

print("\nTotal Customers:", len(df_clusters))
```

```python

# Analyze customer characteristics in each cluster

# Calculate subscription rate
df_clusters['Subscribed'] = (
    df_clusters['y'] == 'yes'
).astype(int)

# Create cluster summary
cluster_summary = df_clusters.groupby('Cluster').agg(
    Customers=('age', 'count'),
    Average_Age=('age', 'mean'),
    Average_Campaign_Contacts=('campaign', 'mean'),
    Subscription_Rate=('Subscribed', 'mean')
)

# Convert subscription rate to percentage
cluster_summary['Subscription_Rate'] *= 100

# Round results
cluster_summary = cluster_summary.round(2)

# Display summary
print("Cluster Characteristics:")
display(cluster_summary)
```

```python
# Visualize subscription rate by cluster

plt.figure(figsize=(8, 5))

sns.barplot(
    x=cluster_summary.index,
    y=cluster_summary['Subscription_Rate'],
    color='steelblue'
)

plt.title('Term Deposit Subscription Rate by Cluster')
plt.xlabel('Cluster')
plt.ylabel('Subscription Rate (%)')
plt.xticks([0, 1], ['Cluster 0', 'Cluster 1'])

# Add percentage labels
for i, value in enumerate(cluster_summary['Subscription_Rate']):
    plt.text(i, value + 0.5, f'{value:.2f}%',
             ha='center')

plt.ylim(0, 28)
plt.tight_layout()
plt.show()
```

```python
# Visualize K-Means clusters using PCA

from sklearn.decomposition import PCA

# Reduce features to two dimensions
pca = PCA(n_components=2, random_state=42)

X_pca = pca.fit_transform(X_cluster_scaled)

# Create a sample for visualization
plot_indices = np.random.RandomState(42).choice(
    len(X_pca),
    size=5000,
    replace=False
)

# Plot customer clusters
plt.figure(figsize=(9, 6))

sns.scatterplot(
    x=X_pca[plot_indices, 0],
    y=X_pca[plot_indices, 1],
    hue=cluster_labels[plot_indices],
    palette='Set1',
    alpha=0.5,
    s=25
)

plt.title('Customer Segmentation Using K-Means and PCA')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster')

plt.tight_layout()
plt.show()
```

```python
# 4.7 Build Logistic Regression Model

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score
from sklearn.metrics import recall_score, f1_score

# Create Logistic Regression pipeline
logistic_model = Pipeline([
    ('preprocessing', mode_preprocessor),
    ('classifier', LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Train the model
logistic_model.fit(X_train_mode, y_train)

# Predict using testing data
logistic_predictions = logistic_model.predict(X_test_mode)

# Evaluate the model
print("Logistic Regression Results:")
print("Accuracy:", round(accuracy_score(
    y_test, logistic_predictions), 4))
print("Precision:", round(precision_score(
    y_test, logistic_predictions), 4))
print("Recall:", round(recall_score(
    y_test, logistic_predictions), 4))
print("F1-score:", round(f1_score(
    y_test, logistic_predictions), 4))
```

```python
# 4.7 Build Random Forest Model

from sklearn.ensemble import RandomForestClassifier

# Create Random Forest pipeline
rf_model = Pipeline([
    ('preprocessing', mode_preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=100,
        max_depth=12,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model
rf_model.fit(X_train_mode, y_train)

# Predict using testing data
rf_predictions = rf_model.predict(X_test_mode)

# Evaluate the model
print("Random Forest Results:")
print("Accuracy:", round(
    accuracy_score(y_test, rf_predictions), 4
))
print("Precision:", round(
    precision_score(y_test, rf_predictions), 4
))
print("Recall:", round(
    recall_score(y_test, rf_predictions), 4
))
print("F1-score:", round(
    f1_score(y_test, rf_predictions), 4
))
```

```python
# Tune Random Forest using RandomizedSearchCV

from sklearn.model_selection import RandomizedSearchCV

# Define parameter combinations
param_grid = {
    'classifier__n_estimators': [100, 150],
    'classifier__max_depth': [8, 12, 16],
    'classifier__min_samples_leaf': [2, 5, 10],
    'classifier__class_weight': [None, 'balanced']
}

# Create tuning pipeline
rf_tuning_pipeline = Pipeline([
    ('preprocessing', mode_preprocessor),
    ('classifier', RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ))
])

# Randomized search with cross-validation
rf_search = RandomizedSearchCV(
    estimator=rf_tuning_pipeline,
    param_distributions=param_grid,
    n_iter=10,
    scoring='f1',
    cv=3,
    random_state=42,
    n_jobs=1
)

# Fit using training data only
rf_search.fit(X_train_mode, y_train)

# Display best parameters
print("Best Parameters:")
print(rf_search.best_params_)

print("\nBest Cross-Validation F1-score:")
print(round(rf_search.best_score_, 4))
```

```python

# Evaluate Tuned Random Forest Model

# Get the best model from RandomizedSearchCV
best_rf_model = rf_search.best_estimator_

# Predict using testing data
tuned_rf_predictions = best_rf_model.predict(X_test_mode)

# Calculate evaluation metrics
print("Tuned Random Forest Results:")

print("Accuracy:", round(
    accuracy_score(y_test, tuned_rf_predictions), 4
))

print("Precision:", round(
    precision_score(y_test, tuned_rf_predictions), 4
))

print("Recall:", round(
    recall_score(y_test, tuned_rf_predictions), 4
))

print("F1-score:", round(
    f1_score(y_test, tuned_rf_predictions), 4
))
```

```python
# Compare classification models

model_results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Random Forest',
        'Tuned Random Forest'
    ],
    'Accuracy': [
        accuracy_score(y_test, logistic_predictions),
        accuracy_score(y_test, rf_predictions),
        accuracy_score(y_test, tuned_rf_predictions)
    ],
    'Precision': [
        precision_score(y_test, logistic_predictions),
        precision_score(y_test, rf_predictions),
        precision_score(y_test, tuned_rf_predictions)
    ],
    'Recall': [
        recall_score(y_test, logistic_predictions),
        recall_score(y_test, rf_predictions),
        recall_score(y_test, tuned_rf_predictions)
    ],
    'F1-score': [
        f1_score(y_test, logistic_predictions),
        f1_score(y_test, rf_predictions),
        f1_score(y_test, tuned_rf_predictions)
    ]
})

# Round results
model_results = model_results.round(4)

# Display comparison
display(model_results)
```

```python
# ROC Curve and AUC Comparison

from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Get predicted probabilities for positive class
logistic_probs = logistic_model.predict_proba(
    X_test_mode
)[:, 1]

rf_probs = rf_model.predict_proba(
    X_test_mode
)[:, 1]

tuned_rf_probs = best_rf_model.predict_proba(
    X_test_mode
)[:, 1]

# Plot ROC curves
plt.figure(figsize=(8, 6))

for name, probs in [
    ('Logistic Regression', logistic_probs),
    ('Random Forest', rf_probs),
    ('Tuned Random Forest', tuned_rf_probs)
]:
    fpr, tpr, thresholds = roc_curve(y_test, probs)
    auc_score = roc_auc_score(y_test, probs)

    plt.plot(
        fpr,
        tpr,
        label=f'{name} (AUC = {auc_score:.3f})'
    )

# Random classifier reference line
plt.plot([0, 1], [0, 1], 'k--', label='Random Classifier')

plt.title('ROC Curve Comparison of Classification Models')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Precision-Recall Curve Comparison

from sklearn.metrics import precision_recall_curve
from sklearn.metrics import average_precision_score

plt.figure(figsize=(8, 6))

for name, probs in [
    ('Logistic Regression', logistic_probs),
    ('Random Forest', rf_probs),
    ('Tuned Random Forest', tuned_rf_probs)
]:

    precision, recall, thresholds = precision_recall_curve(
        y_test, probs
    )

    ap_score = average_precision_score(y_test, probs)

    plt.plot(
        recall,
        precision,
        label=f'{name} (AP = {ap_score:.3f})'
    )

# Baseline: proportion of positive class
baseline = y_test.mean()

plt.axhline(
    baseline,
    linestyle='--',
    color='gray',
    label=f'Baseline = {baseline:.3f}'
)

plt.title('Precision-Recall Curve Comparison')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
